```python
import os
import requests

URL = "https://api.gptzero.me/v2/predict/files"
HEADERS = {
    "Content-Type": "multipart/form-data",
    "Accept": "application/json"
}


def check_files_with_gptzero(file_paths):
    files = [("files", open(file, "rb")) for file in file_paths]
    response = requests.post(URL, files=files, headers=HEADERS)
    for _, file in files:
        file.close()
    return response.json()


def summarize_file(content):
    sentences = content.split(".")
    return ". ".join(sentences[:3]) + "..."


DOCUMENTS_FOLDER = "./documents"

txt_files = [os.path.join(DOCUMENTS_FOLDER, file_name) for file_name in os.listdir(DOCUMENTS_FOLDER) if file_name.endswith(".txt")]
result = check_files_with_gptzero(txt_files)
print(result)
for file_path, document in zip(txt_files, result.get("documents", [])):
    with open(file_path, "r", encoding="utf-8") as file:
        content = file.read()
        summary = summarize_file(content)
        print(f"{os.path.basename(file_path)}")
        print(f"Class: {document.get('predicted_class')}")
        print(f"Résumé: {summary}")
```
